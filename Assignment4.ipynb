### Q1: Basic Dictionary Creation

```python
# Student ki details ka dictionary bana kar values print kar rahe hain
student = {
    "name": "Hassan",
    "age": 21,
    "course": "Data Analytics with Python",
    "city": "Karachi"
}

for key, value in student.items():
    print(f"{key}: {value}")
```

### Q2: Adding & Updating Values

```python
# Dictionary mein color add kar rahe hain aur year update kar rahe hain
car = {
    "brand": "Toyota",
    "model": "Corolla",
    "year": 2018
}

car["color"] = "White"
car["year"] = 2022

print("Updated Car Dictionary:", car)
```

### Q3: Loop Through Dictionary

```python
# Loop ki madad se dictionary ki keys aur values print kar rahe hain
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}

for k, v in person.items():
    print(f"Key: {k} --> Value: {v}")
```

### Q4: Count Frequency of Numbers

```python
# Numbers ki list se frequency dictionary bana rahe hain
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
freq_dict = {}

for num in numbers:
    if num in freq_dict:
        freq_dict[num] += 1
    else:
        freq_dict[num] = 1

print("Frequency Dictionary:", freq_dict)
```

### Q5: Word Counter from Sentence

```python
# Sentence mein words kitni baar aaye hain uska count nikal rahe hain
sentence = input("Enter a sentence to count words: ")
words = sentence.split()
word_count = {}

for w in words:
    w_lower = w.lower()
    word_count[w_lower] = word_count.get(w_lower, 0) + 1

print("Word Frequencies:", word_count)
```

### Q6: Student Marks & Average

```python
# Subjects ke marks ka total aur average nikal rahe hain
marks = {
    "Math": 80, 
    "English": 70, 
    "Physics": 90
}

total_marks = sum(marks.values())
average_marks = total_marks / len(marks)

print(f"Total Marks: {total_marks}")
print(f"Average Marks: {average_marks}")
```

### Q7: Highest Scoring Student

```python
# Dictionary mein se sab se zyada marks wale student ko dhoond rahe hain
students_scores = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_student = max(students_scores, key=students_scores.get)
print(f"The highest scorer is {highest_student} with {students_scores[highest_student]} marks.")
```

### Q8: Merge Dictionaries

```python
# Do dictionaries ko apas mein merge kar rahe hain
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged_dict = dict1 | dict2
print("Merged Dictionary:", merged_dict)
```
