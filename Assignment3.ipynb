### Question 1: Print alternate elements of a list

```python
# User se list input le rahe hain aur alternate elements print kar rahe hain
user_input = input("Enter elements separated by space: ")
my_list = user_input.split()

print("Alternate elements are:")
for i in range(0, len(my_list), 2):
    print(my_list[i])
```

### Question 2: Reverse list content without using reverse() method

```python
# User se list le kar bina reverse() function ke usay ulta kar rahe hain
user_input = input("Enter elements separated by space: ")
my_list = user_input.split()

reversed_list = []
for i in range(len(my_list) - 1, -1, -1):
    reversed_list.append(my_list[i])

print("Reversed list:", reversed_list)
```

### Question 3: Find largest number without using max()

```python
# Keyboard se numbers le kar sab se bara number dhoond rahe hain bina max() ke
numbers_input = input("Enter numbers separated by space: ")
numbers = [float(x) for x in numbers_input.split()]

largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num

print("The largest number is:", largest)
```

### Question 4: Rotate list elements

```python
# List ke elements ko right side par aik step shift kar rahe hain
user_input = input("Enter elements separated by space: ")
my_list = user_input.split()

if len(my_list) > 1:
    last_element = my_list.pop()
    my_list.insert(0, last_element)

print("Rotated list:", my_list)
```

### Question 5: Delete a given word from a string

```python
# User se sentence aur delete karne wala word le kar usay remove kar rahe hain
sentence = input("Enter a sentence: ")
word_to_delete = input("Enter the word you want to delete: ")

words_list = sentence.split()
new_words = []
for w in words_list:
    if w.lower() != word_to_delete.lower():
        new_words.append(w)

final_sentence = " ".join(new_words)
print("Updated sentence:", final_sentence)
```

### Question 6: Convert mm/dd/yyyy date format

```python
# Date ko mm/dd/yyyy se Month Day, Year format mein convert kar rahe hain
date_input = input("Enter date in mm/dd/yyyy format (e.g., 03/12/2021): ")

parts = date_input.split("/")
month_num = parts[0]
day = parts[1]
year = parts[2]

months_dict = {
    "01": "January", "02": "February", "03": "March", "04": "April",
    "05": "May", "06": "June", "07": "July", "08": "August",
    "09": "September", "10": "October", "11": "November", "12": "December"
}

month_name = months_dict.get(month_num, "Invalid Month")
print(f"Formatted Date: {month_name} {day}, {year}")
```

### Question 8: Sum of each row of a matrix

```python
# Matrix ki har row ka sum nikal rahe hain
matrix = [
    [2, 11, 7],
    [12, 5, 2],
    [9, 15, 8]
]

row_num = 1
for row in matrix:
    row_sum = sum(row)
    print(f"Sum of row {row_num} = {row_sum}")
    row_num += 1
```

### Question 9: Add two matrices

```python
# Do matrices ko apas mein add kar rahe hain
matrix1 = [[1, 2], [3, 4]]
matrix2 = [[5, 6], [7, 8]]

result_matrix = [[0, 0], [0, 0]]

for i in range(len(matrix1)):
    for j in range(len(matrix1[0])):
        result_matrix[i][j] = matrix1[i][j] + matrix2[i][j]

print("Sum of matrices:")
for r in result_matrix:
    print(r)
```

### Question 10: Multiply two matrices

```python
# Do matrices ko multiply kar rahe hain
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]

mul_result = [[0, 0], [0, 0]]

for i in range(len(A)):
    for j in range(len(B[0])):
        for k in range(len(B)):
            mul_result[i][j] += A[i][k] * B[k][j]

print("Multiplication of matrices:")
for r in mul_result:
    print(r)
```
