Question 1Create a 1D NumPy array of integers from 10 to 50 (inclusive, step 5). Print the array, its shape, dtype and number of dimensions.   

```python
import numpy as np

# 1D array creation (10 to 50 inclusive with step 5)
arr = np.arange(10, 51, 5)

print("Array:", arr)
print("Shape:", arr.shape)
print("Data Type:", arr.dtype)
print("Number of Dimensions:", arr.ndim)
```

Question 2
Create a 3x4 array of zeros, a 3x4 array of ones, and a 4x4 identity matrix. Then create an array of 8 evenly spaced numbers between 0 and 1 using linspace.

```python
# 3x4 Zeros array
zeros_arr = np.zeros((3, 4))
print("Zeros Array (3x4):\n", zeros_arr)

# 3x4 Ones array
ones_arr = np.ones((3, 4))
print("\nOnes Array (3x4):\n", ones_arr)

# 4x4 Identity matrix
identity_mat = np.eye(4)
print("\nIdentity Matrix (4x4):\n", identity_mat)

# 8 evenly spaced numbers between 0 and 1
linspace_arr = np.linspace(0, 1, 8)
print("\nLinspace Array (8 numbers):\n", linspace_arr)
```

Question 3
Create a 1D array of numbers from 1 to 24 and reshape it into a 2D array of shape (4, 6). Then reshape it into a 3D array of shape (2, 3, 4) and print the shape of each.

```python
# 1D array from 1 to 24
arr1d = np.arange(1, 25)
print("1D Array Shape:", arr1d.shape)

# Reshape to 2D (4, 6)
arr2d = arr1d.reshape(4, 6)
print("2D Array Shape:", arr2d.shape)

# Reshape to 3D (2, 3, 4)
arr3d = arr1d.reshape(2, 3, 4)
print("3D Array Shape:", arr3d.shape)
```

Question 4
Given arr = np.array([12, 45, 7, 23, 56, 89, 34, 21]), print: the first 3 elements, the last 3 elements, every alternate element, and the array in reverse order.

```python
arr = np.array([12, 45, 7, 23, 56, 89, 34, 21])

print("First 3 elements:", arr[:3])
print("Last 3 elements:", arr[-3:])
print("Every alternate element:", arr[::2])
print("Reversed array:", arr[::-1])
```

Question 5
Create a 2D array of shape (5, 5) with values 1 to 25. Extract (a) the second row, (b) the third column, (c) the 3x3 sub-matrix from the center.

```python
# 5x5 array with values 1 to 25
arr2d = np.arange(1, 26).reshape(5, 5)
print("Original 5x5 Array:\n", arr2d)

# (a) Second row (index 1)
print("\nSecond Row:", arr2d[1, :])

# (b) Third column (index 2)
print("Third Column:", arr2d[:, 2])

# (c) 3x3 Center Sub-matrix
print("3x3 Center Sub-matrix:\n", arr2d[1:4, 1:4])
```

Question 6
Given a = np.array([10, 20, 30, 40]) and b = np.array([1, 2, 3, 4]), perform addition, subtraction, multiplication, division and power (a ** 2). Also compute the dot product of a and b.

```python
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power (a ** 2):", a ** 2)

# Dot product
print("Dot Product:", np.dot(a, b))
```

Question 7
Given arr = np.array([[4, 8, 2], [9, 1, 7], [5, 6, 3]]), find the sum, mean, min, max and standard deviation of the whole array, and then the sum along rows (axis=1) and along columns (axis=0).

```python
arr = np.array([[4, 8, 2], [9, 1, 7], [5, 6, 3]])

# Whole array stats
print("Total Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Min:", np.min(arr))
print("Max:", np.max(arr))
print("Standard Deviation:", np.std(arr))

# Sum along rows and columns
print("Sum along rows (axis=1):", np.sum(arr, axis=1))
print("Sum along columns (axis=0):", np.sum(arr, axis=0))
```

Question 8
Create an array of 20 random integers between 1 and 100. Using boolean masking, print (a) all elements greater than 50, (b) all even elements, (c) count of elements divisible by 5.

```python
np.random.seed(42)  # For reproducible random values
random_arr = np.random.randint(1, 101, size=20)
print("Random Array:", random_arr)

# (a) Elements > 50
print("Elements > 50:", random_arr[random_arr > 50])

# (b) Even elements
print("Even elements:", random_arr[random_arr % 2 == 0])

# (c) Count of elements divisible by 5
div_by_5_count = np.sum(random_arr % 5 == 0)
print("Count of elements divisible by 5:", div_by_5_count)
```

Question 9
Use np.where to replace all negative values in np.array([5, -3, 8, -1, 0, -7, 4]) with 0. Also find the index of the maximum and minimum value using argmax and argmin.

```python
arr = np.array([5, -3, 8, -1, 0, -7, 4])

# Replace negative numbers with 0
replaced_arr = np.where(arr < 0, 0, arr)
print("Replaced Array:", replaced_arr)

# Argmax & Argmin on original array
print("Index of Max value:", np.argmax(arr))
print("Index of Min value:", np.argmin(arr))
```

Question 10
Create two 2x3 arrays and join them vertically (vstack) and horizontally (hstack). Then sort the array np.array([[9, 3, 6], [2, 8, 1]]) along each row and flatten the result into 1D.

```python
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr2 = np.array([[7, 8, 9], [10, 11, 12]])

# Vertical and Horizontal stack
print("Vstack:\n", np.vstack((arr1, arr2)))
print("Hstack:\n", np.hstack((arr1, arr2)))

# Sorting along each row and flattening
sample_arr = np.array([[9, 3, 6], [2, 8, 1]])
sorted_rows = np.sort(sample_arr, axis=1)
flattened_arr = sorted_rows.flatten()

print("\nSorted along rows:\n", sorted_rows)
print("Flattened 1D Array:", flattened_arr)
```
