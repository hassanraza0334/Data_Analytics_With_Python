Question 1
Create a DataFrame from a dictionary with columns Name, Age, City and Salary for 6 employees. Print the first 3 rows, the shape, the column names and the data types.

```python
import pandas as pd

# Creating dictionary for 6 employees
data = {
    'Name': ['Ali', 'Usman', 'Sara', 'Zain', 'Ayesha', 'Hamza'],
    'Age': [25, 32, 28, 40, 22, 35],
    'City': ['Lahore', 'Delhi', 'Karachi', 'Delhi', 'Islamabad', 'Lahore'],
    'Salary': [45000, 55000, 38000, 62000, 30000, 48000]
}

df = pd.DataFrame(data)

print("First 3 rows:\n", df.head(3))
print("\nShape of DataFrame:", df.shape)
print("\nColumn Names:", df.columns.tolist())
print("\nData Types:\n", df.dtypes)
```

Question 2
Create a Series of 5 numbers with custom index labels ['a','b','c','d','e']. Print the value at label 'c', the values at positions 1 to 3, and the sum and mean of the Series.

```python
# Series creation with custom index
numbers = [10, 25, 30, 45, 50]
labels = ['a', 'b', 'c', 'd', 'e']
s = pd.Series(numbers, index=labels)

print("Series:\n", s)
print("\nValue at label 'c':", s['c'])
print("\nValues at positions 1 to 3:\n", s.iloc[1:4])
print("\nSum of Series:", s.sum())
print("Mean of Series:", s.mean())
```

Question 6
Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
# Adding Bonus and Total columns
df['Bonus'] = df['Salary'] * 0.10
df['Total'] = df['Salary'] + df['Bonus']

# Renaming City to Location
df = df.rename(columns={'City': 'Location'})

# Dropping Bonus column
df = df.drop(columns=['Bonus'])

print("Updated DataFrame:\n", df)
```

Question 7
Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
import numpy as np

# DataFrame with NaNs
data_with_nan = {
    'Name': ['Ali', 'Usman', np.nan, 'Zain'],
    'Age': [25, np.nan, 28, 40],
    'Salary': [45000, 50000, 38000, np.nan]
}

df_nan = pd.DataFrame(data_with_nan)
print("Original DataFrame with NaNs:\n", df_nan)

# Missing values count per column
print("\nMissing values count:\n", df_nan.isnull().sum())

# Fill numeric NaNs with column mean
df_nan['Age'] = df_nan['Age'].fillna(df_nan['Age'].mean())
df_nan['Salary'] = df_nan['Salary'].fillna(df_nan['Salary'].mean())

# Drop remaining rows with NaNs (e.g. Name column)
df_clean = df_nan.dropna()

print("\nFinal Clean DataFrame:\n", df_clean)
```

Question 8
Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
# Sort by Salary descending
df_sorted_salary = df.sort_values(by='Salary', ascending=False)
print("Sorted by Salary (Desc):\n", df_sorted_salary)

# Sort by Age ascending
df_sorted_age = df.sort_values(by='Age', ascending=True)
print("\nSorted by Age (Asc):\n", df_sorted_age)

# Top 3 highest-paid employees
top_3_paid = df.nlargest(3, 'Salary')
print("\nTop 3 Highest-Paid Employees:\n", top_3_paid)
```

Question 9
Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
# Groupby Location (since City was renamed in Q6) with agg()
grouped_data = df.groupby('Location').agg(
    Avg_Salary=('Salary', 'mean'),
    Employee_Count=('Salary', 'count'),
    Max_Salary=('Salary', 'max')
)

print("Aggregated Groupby Result:\n", grouped_data)
```

Question 10
Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
# 1. Creating DataFrames
emp = pd.DataFrame({
    'EmpID': [1, 2, 3, 4],
    'Name': ['Ali', 'Sara', 'Usman', 'Zain'],
    'DeptID': [101, 102, 101, 103]
})

dept = pd.DataFrame({
    'DeptID': [101, 102, 104],
    'DeptName': ['IT', 'HR', 'Finance']
})

# Inner Join
inner_merge = pd.merge(emp, dept, on='DeptID', how='inner')
print("Inner Join:\n", inner_merge)

# Left Join
left_merge = pd.merge(emp, dept, on='DeptID', how='left')
print("\nLeft Join:\n", left_merge)

# 2. Concatenation and removing duplicates
emp2 = pd.DataFrame({
    'EmpID': [3, 5],
    'Name': ['Usman', 'Hamza'],
    'DeptID': [101, 102]
})

concatenated = pd.concat([emp, emp2], ignore_index=True)
print("\nConcatenated DataFrame:\n", concatenated)

# Remove duplicates
unique_df = concatenated.drop_duplicates()
print("\nAfter Removing Duplicates:\n", unique_df)
```
