### Question 1: Working with Lists, Loops, and Conditions

```python
# List banai hai jismein 8 integers hain
numbers = [12, 5, 8, 20, 15, 3, 10, 22]

print("--- All Numbers in List ---")
for num in numbers:
    print(num)

# Saare numbers ka sum nikal rahe hain
total_sum = sum(numbers)
print("\nSum of all numbers:", total_sum)

print("\n--- Only Even Numbers ---")
for num in numbers:
    if num % 2 == 0:
        print(num)
```

### Question 2: Dictionary Basics and Student Average Marks

```python
# Student ka dictionary banaya hai jismein marks ki list hai
student = {
    "name": "Ali",
    "age": 21,
    "course": "Data Analytics with Python",
    "marks": [75, 82, 68]
}

print("--- Student Details ---")
for key, value in student.items():
    print(f"{key}: {value}")

# Loop ki madad se marks ka average nikal rahe hain
total_marks = 0
count = 0
for m in student["marks"]:
    total_marks += m
    count += 1

average_marks = total_marks / count
print(f"\nAverage Marks: {average_marks}")

# Condition check kar rahe hain ke pass hai ya fail
if average_marks >= 50:
    print("Status: Passed")
else:
    print("Status: Failed")
```

### Question 3: List of Dictionaries (Employees Data)

```python
# 3 employees ki details ka list of dictionaries banaya hai
employees = [
    {"name": "Ahmed", "department": "IT", "salary": 45000},
    {"name": "Sara", "department": "HR", "salary": 60000},
    {"name": "Zain", "department": "Finance", "salary": 55000}
]

print("--- Employees Details ---")
total_expense = 0
highest_salary = employees[0]["salary"]
highest_paid_employee = employees[0]["name"]

for emp in employees:
    print(f"Name: {emp['name']}, Dept: {emp['department']}, Salary: {emp['salary']}")
    total_expense += emp["salary"]
    
    # Highest salary find karne ki logic
    if emp["salary"] > highest_salary:
        highest_salary = emp["salary"]
        highest_paid_employee = emp["name"]

print(f"\nTotal Salary Expense: {total_expense}")
print(f"Highest Paid Employee: {highest_paid_employee} ({highest_salary})")
```

### Question 4: While Loop with User Input (-1 to Stop)

```python
# While loop ke zariye user se numbers le rahe hain jab tak wo -1 na enter kare
user_numbers = []
print("Enter numbers (Type -1 to stop):")

while True:
    val = float(input("Enter number: "))
    if val == -1:
        break
    user_numbers.append(val)

print("\nComplete List:", user_numbers)

if len(user_numbers) > 0:
    print("Largest number:", max(user_numbers))
    print("Smallest number:", min(user_numbers))
else:
    print("No numbers were entered.")
```

### Question 5: Word Frequency Counter using Dictionary

```python
# User se sentence le kar words ki frequency count kar rahe hain
sentence = input("Enter a sentence: ")
words_list = sentence.split()

word_freq = {}
for word in words_list:
    w_lower = word.lower()
    word_freq[w_lower] = word_freq.get(w_lower, 0) + 1

print("\nWord Frequencies:")
for w, count in word_freq.items():
    print(f"{w}: {count}")
```

### Question 6: Nested Loops – Multiplication Table (1 to 5)

```python
# Nested loops ki madad se 1 se 5 tak multiplication table print kar rahe hain
print("--- Multiplication Table (1 to 5) ---")
for i in range(1, 6):
    row_output = ""
    for j in range(1, 6):
        row_output += f"{i * j}\t"
    print(row_output)
```
