1) A company decided to give bonus of 5% to employee if his/her year of service is more than 5 years. Ask user for their salary and year of service and print the net bonus amount.

```python
# User se salary aur years of service input le rahay hain
salary = float(input("Enter your salary: "))
years_of_service = float(input("Enter your years of service: "))

# Condition check kar rahay hain ke service 5 saal se zyada hai ya nahi
if years_of_service > 5:
    # Agar 5 saal se zyada hai toh 5% bonus calculate hoga
    bonus = salary * 0.05
    print(f"Congratulations! You got a 5% bonus. Net bonus amount is: {bonus}")
else:
    # Agar service 5 saal ya us se kam hai toh koi bonus nahi milay ga
    print("Sorry, you are not eligible for a bonus (service must be greater than 5 years).")
```

6) Write a program to calculate the electricity bill (accept number of unit from user) according to the following criteria:
- upto 100 units: no charge
- Next 200 units: Rs 5 per unit
- After 200 units: Rs 10 per unit

```python
# User se electricity units input le rahay hain
units = int(input("Enter consumed electricity units: "))
bill = 0

# Bill calculate karne ki logic likh rahay hain
if units <= 100:
    bill = 0
elif units <= 300:
    # Pehle 100 par koi charge nahi, baqi units par Rs 5 charge hoga
    bill = (units - 100) * 5
else:
    # Pehle 100 free, agle 200 units par Rs 5, aur 300 se upar wale units par Rs 10
    bill = (200 * 5) + (units - 300) * 10

print(f"Total electricity bill amount is: Rs. {bill}")
```

11) A shop will give discount of 10% if the cost of purchased quantity is more than 1000. Ask user for quantity. Suppose, one unit will cost 100. Judge and print total cost for user.

```python
# User se purchased quantity input le rahay hain
quantity = int(input("Enter quantity purchased: "))

# 1 unit ki cost 100 hai toh total cost calculate karenge
total_cost = quantity * 100

# Check karenge ke total cost 1000 se zyada hai ya nahi taake discount apply ho sakay
if total_cost > 1000:
    discount = total_cost * 0.10
    total_cost = total_cost - discount
    print(f"You got a 10% discount! Your final total cost is: Rs. {total_cost}")
else:
    print(f"No discount applied. Your total cost is: Rs. {total_cost}")
```

12) A school has following rules for grading system: Below 25 - F, 25 to 45 - E, 45 to 50 - D, 50 to 60 - C, 60 to 80 - B, Above 80 - A. Ask user to enter marks and print the corresponding grade.

```python
# User se marks input le rahay hain
marks = float(input("Enter your marks: "))

# If-elif ladder use kar ke appropriate grade assign kar rahay hain
if marks < 25:
    print("Grade: F")
elif marks >= 25 and marks < 45:
    print("Grade: E")
elif marks >= 45 and marks < 50:
    print("Grade: D")
elif marks >= 50 and marks < 60:
    print("Grade: C")
elif marks >= 60 and marks < 80:
    print("Grade: B")
elif marks >= 80 and marks <= 100:
    print("Grade: A")
else:
    print("Invalid marks entered!")
```

13) Take input of age of 3 people by user and determine oldest and youngest among them.

```python
# Teen logon ki age input le rahay hain
age1 = int(input("Enter age of first person: "))
age2 = int(input("Enter age of second person: "))
age3 = int(input("Enter age of third person: "))

# Oldest find karne ki condition
if age1 >= age2 and age1 >= age3:
    oldest = age1
elif age2 >= age1 and age2 >= age3:
    oldest = age2
else:
    oldest = age3

# Youngest find karne ki condition
if age1 <= age2 and age1 <= age3:
    youngest = age1
elif age2 <= age1 and age2 <= age3:
    youngest = age2
else:
    youngest = age3

print(f"The oldest person is {oldest} years old.")
print(f"The youngest person is {youngest} years old.")
```

14) A student will not be allowed to sit in exam if his/her attendance is less than 75%. Take number of classes held and attended, and print percentage and allowed status.

```python
# Classes held aur attended ka data input le rahay hain
classes_held = int(input("Enter total number of classes held: "))
classes_attended = int(input("Enter total number of classes attended: "))

# Attendance percentage calculate kar rahay hain
attendance_percentage = (classes_attended / classes_held) * 100
print(f"Your attendance percentage is: {attendance_percentage:.2f}%")

# Check kar rahay hain ke percentage 75% ya us se zyada hai ya nahi
if attendance_percentage >= 75:
    print("You are allowed to sit in the exam.")
else:
    print("You are NOT allowed to sit in the exam (attendance is less than 75%).")
```

16) Write a program to check if a year is leap year or not. If a year is divisible by 4 then it is leap year but if the year is century year like 2000, 1900, 2100 then it must be divisible by 400.

```python
# User se year input le rahay hain
year = int(input("Enter a year: "))

# Century year aur normal leap year ki conditions check kar rahay hain
if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(f"{year} is a Leap Year.")
else:
    print(f"{year} is not a Leap Year.")
```
